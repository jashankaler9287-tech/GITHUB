```python
#Ques1:
import numpy as np
np.random.seed(4)
x=np.random.randint(1,100,(10,10))

print("Original Array:")
print(x)

A=x[0:5,0:7]
B=x[0:7,0:5]
C=x[5:10,3:10]

print("A:")
print(A)

print("B:")
print(B)

print("C:")
print(C)

print("Sum of A and C:")
print(A + C)

print("Multiplication of A and B:")
print(A @ B)
```

```python
#Ques2:
np.random.seed(5)
x=np.random.randint(1,100,(10,10))

print("Original Array:")
print(x)

#2D reshapes
print("\n1 * 100:")
print(x.reshape(1,100))
print("\n2 * 50:")
print(x.reshape(2,50))
print("\n4 * 25:")
print(x.reshape(4,25))
print("\n5 * 20:")
print(x.reshape(5,20))
print("\n10 * 10:")
print(x.reshape(10,10))
print("\n20 * 5:")
print(x.reshape(20,5))
print("\n25 * 4:")
print(x.reshape(25,4))
print("\n50 * 2:")
print(x.reshape(50,2))
print("\n100 * 1:")
print(x.reshape(100,1))

#3D reshapes
print("\n2 * 5 * 10:")
print(x.reshape(2,5,10))
print("\n2 * 10 * 5:")
print(x.reshape(2,10,5))
print("\n5 * 2 * 10:")
print(x.reshape(5,2,10))
print("\n5 * 10 * 2:")
print(x.reshape(5,10,2))
print("\n10 * 2 * 5:")
print(x.reshape(10,2,5))
print("\n10 * 5 * 2:")
print(x.reshape(10,5,2))
```

```python
#Ques3:
np.random.seed(42)
x=np.random.randint(10,99,(6,6))

print("Original Array:")
print(x)

print("Sum of each row:")
print(x.sum(axis=1))

print("Sum of each column:")
print(x.sum(axis=0))

print("Mean of each row:")
print(np.round((x.mean(axis=1)),2))

print("Maximum value in each row:")
print(x.max(axis=1))

print("Column index of max value in each row:")
print(x.argmax(axis=1))

print("Array converted to float datatype:")
y=x.astype('float')
print(y)
```

```python
#Ques4:
x=np.arange(1,21)
print("Arranged Array:")
print(x)

x=x.reshape(4,5)
print("Reshaped Array:")
print(x)

sub=x[1:3,2:5]
print("Extracted subarray:")
print(sub)

variance=np.var(sub)
std=np.std(sub)

print("Variance:",variance)
print("Standard Deviation:",std)
```

```python
#Ques5:
np.random.seed(21)
x=np.random.uniform(50,150,8)
print("Raw Array:")
print(x)

print("Values rounded to 2 decimal places:")
rounded=np.round(x,2)
print(rounded)

zeros=np.zeros((2,4))
ones=np.ones((2,4))
print("\nZeros:")
print(zeros)
print("\nOnes:")
print(ones)

reshaped=rounded.reshape(2,4)
print("\nReshaped Array:")
print(reshaped)

result = reshaped+10
print("\nAfter adding 10:")
print(result)
```
